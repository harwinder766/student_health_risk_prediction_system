# Applying 5 fold Cross Validation

```python
le = LabelEncoder()
y = le.fit_transform(train[TARGET])

print("Class encoding: ",dict(zip(le.classes_,range(len(le.classes_)))))

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
FOLD_SPLITS = list(skf.split(train,y))

oof_probs = np.zeros((len(train),N_CLASSES))
test_probs = np.zeros((len(test),N_CLASSES))

fold_scores = []
models = []
fi_list = []
encoding_infos = []

cat_cols = [col for col in train.select_dtypes(include = 'object').columns if col != TARGET]

```

```python
LGB_PARAMS = dict(
    objective="multiclass",
    num_class=N_CLASSES,

    learning_rate=0.03,
    n_estimators=6000,

    num_leaves=31,
    max_depth=-1,

    subsample=0.8,
    colsample_bytree=0.8,

    reg_alpha=0.0,
    reg_lambda=1.0,

    random_state=SEED,

    class_weight="balanced",

    verbosity=-1,
)
```

```python
train.select_dtypes(
    include=["object", ]
).columns.tolist()
```

```python
import lightgbm as lgb

for fold, (tr_idx, val_idx) in enumerate(FOLD_SPLITS, 1):

    print(f"\nFold {fold}")

    (
        train_fold,
        valid_fold,
        test_fold,
        encoding_info
    ) = prepare_fold(
        train_df=train,
        test_df=test,
        train_idx=tr_idx,
        valid_idx=val_idx,
        cat_cols=cat_cols,
        target=TARGET
    )

    encoding_infos.append(encoding_info)

    FEATURES_FOLD = [
        col for col in train_fold.columns
        if col != TARGET
    ]

    # Make categorical columns category dtype
    for c in cat_cols:

        categories = train_fold[c].astype(str).unique()

        train_fold[c] = pd.Categorical(
            train_fold[c].astype(str),
            categories=categories
        )

        valid_fold[c] = pd.Categorical(
            valid_fold[c].astype(str),
            categories=categories
        )

        test_fold[c] = pd.Categorical(
            test_fold[c].astype(str),
            categories=categories
        )

    y_train = y[tr_idx]
    y_valid = y[val_idx]

    model = LGBMClassifier(**LGB_PARAMS)

    model.fit(
        train_fold[FEATURES_FOLD],
        y_train,

        eval_set=[
            (valid_fold[FEATURES_FOLD], y_valid)
        ],

        eval_metric="multi_logloss",

        callbacks=[
            lgb.early_stopping(200, verbose=False),
            lgb.log_evaluation(0)
        ]
    )

    val_pred = model.predict_proba(
        valid_fold[FEATURES_FOLD]
    )

    test_pred = model.predict_proba(
        test_fold[FEATURES_FOLD]
    )

    oof_probs[val_idx] = val_pred
    test_probs += test_pred / N_FOLDS

    score = balanced_accuracy_score(
        y_valid,
        val_pred.argmax(axis=1)
    )

    fold_scores.append(score)
    models.append(model)

    fi_list.append(
        pd.Series(
            model.feature_importances_,
            index=FEATURES_FOLD,
            name=f"fold{fold}"
        )
    )

    print(
        f"Fold {fold}: "
        f"balanced_acc={score:.5f} "
        f"(best_iter={model.best_iteration_})"
    )
```

```python
raw_oof_score = balanced_accuracy_score(y,oof_probs.argmax(axis=1))

print(f"\nRaw OOF (argmax): "f"{raw_oof_score:.5f}")

print(
    f"CV mean: "
    f"{np.mean(fold_scores):.5f} "
    f"+/- {np.std(fold_scores):.5f}"
)

np.save("../models/oof_lightgbm.npy",oof_probs)

np.save("../models/test_lightgbm.npy",test_probs)
```

**Note**: Segment-specific decision-rule optimization increased OOF balanced accuracy from 0.94797 to approximately 0.9480. However, this improvement did not generalize to the Kaggle test set, where the leaderboard score decreased to approximately 0.947. Therefore, segmented decision optimization was rejected from the final pipeline.

# Conclusion
The baseline 5-fold Lightgbm model achieved an OOF balanced accuracy of 0.93681 using the standard argmax decision rule. Optimizing class-specific decision weights improved the OOF score slightly to 0.94797, indicating that the default argmax rule is not perfectly aligned with the balanced-accuracy objective.

A segmented decision rule was also tested by learning separate class weights for samples with missing and available stress information. This produced an OOF score of approximately 0.94980. Although this is the best OOF score among the tested decision rules, the gain over global optimization is very small, so there is limited evidence that the additional complexity of segmentation provides a meaningful improvement.

Therefore, global decision-rule optimization is preferred for the current pipeline.
